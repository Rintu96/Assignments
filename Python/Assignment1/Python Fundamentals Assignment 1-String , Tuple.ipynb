### 3. String Methods: 

```python
strM='Python beginner tutorial'
strM
```

```python
#Convert the sentence to uppercase
print(strM.upper())
```

```python
#Convert the sentence to lowercase
print(strM.lower())
```

```python
#return the sentence to the original input form
print(strM.capitalize())
```

```python
# total number of occurrences of character ‘t’ 
num=strM.count('t')
print(f'"t" occurs {num} times')
```

```python
#Replace all occurrences of “Python” with “Machine Learning” 
print("Replaced string:", strM.replace("Python", "Machine Learning"))
```

## Tuples (Creation, Modification and Access) :  

```python
t1=(10, 20, 30)
t2=(40, 50, 60)

#Concatenate the two tuples and store it in “t_combine” 
t_combine=t1+t2
print('Combined tuple:',t_combine)
```

```python
#Repeat the elements of “t_combine” 3 times 
print('Repeated tuple:',t_combine*3)
```

```python
#Access the 3rd element from “t_combine”
print('Third element in tuple is:',t_combine[2])
```

```python
#Access the first three elements from “t_combine”  
print('First 3 elements in tuple are:',t_combine[:3])
```

```python
#Access the last three elements from “t_combine” 
print('Last 3 elements in tuple are:',t_combine[-3:])
```
