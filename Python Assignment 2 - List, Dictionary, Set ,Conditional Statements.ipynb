## List (Creation, Modification and Access):

### 1. List Creation:

```python
# a. Create a list named age_list with five integer elements.
age_list=[24, 25, 26, 27, 28]
print(age_list)
```

```python
# Create a list named name_list with five string elements.
name_list=["Alin","Bini","Jesu","Teena","Nate"]
print(name_list)
```

### 2. List Operations / Modifications:

```python
# a. Append the string "Yazhini" to name_list.
name_list.append("Yazhini")
print(name_list)
```

```python
# b. Insert the element 30 at index 2 in age_list.
age_list.insert(2,30)
print(age_list)
```

```python
# c. Remove the string "Yazhini" from name_list.
name_list.remove("Yazhini")
print(name_list)
```

```python
# d. Pop the last element from age_list.
age_list.pop()
print(age_list)
```

```python
# Extend the age_list with additional ages [29, 30, 26].
age_list.extend([29,30,26])
print(age_list)
```

```python
# f. Sort age_list in descending order.
age_list.sort(reverse=True)
print(age_list)
```

```python
# g. Find Max age, Min age and sum of all ages from age_list.
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

### 3. Accessing List Elements:

```python
# a.Print the first element of name_list.
print(name_list[0])
```

```python
# b.Print the last element of name_list.
print(name_list[-1])
```

```python
# c.Print the elements from index 2 to index 4 in name_list.
print(name_list[2:5])
```

```python
# d. Print the elements of name_list in reverse order.
print(name_list[::-1])
```

## Dictionary (Creation, Modification and Access):

```python
# a. Create a dictionary named student_marks that maps the names of five students to their marks (use scale of from 0 to 100).
student_marks={"Alan":70,
               "Ben":85,
               "Cyan":68,
               "Dani":90,
               "Evin":59}
print(student_marks)
```

```python
# b. Access and print the mark of a specific student, of your choice.
print(student_marks["Cyan"])
```

```python
# c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary.
student_marks.update({"Janani":80})
print(student_marks)
```

```python
# d. Update the mark of any one older student to 82.
student_marks["Evin"]=82
print(student_marks)
```

```python
# e. Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.
print("Student names:", student_marks.keys())
print("Student marks:", student_marks.values())
print("Student info:", student_marks.items())
```

## Sets (Operations):

```python
# a. Create a set called my_set with following values:['a','e','i','o','u','a','a','i']
my_set={'a','e','i','o','u','a','a','i'}
print(my_set)
# Explanation: A set stores unique elements only. Duplicate values such as 'a' and 'i' are automatically removed. 
# Sets are unordered, so their display order will not be same as input.
```

```python
# b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.
my_set[4]='s'
```

```python
# Explanation: Sets do not support indexing or changing elements by position. If we want to add 's', we can do:
my_set.add('s')
print(my_set)
```

```python
#c. Create two sets:set1 with values: {1, 3, 5, 7, 9},set2 with values: {2, 3, 5, 8, 10}
set1={1, 3, 5, 7, 9}
set2={2, 3, 5, 8, 10}
print("set1:",set1,"set2:",set2)
```

```python
# d. Compute and print the union and intersection of set1 and set2.
print("Union:",set1|set2)
print("Intersection:",set1&set2)
```

### Performance Category Program:

```python
score=float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent work! Keep it up.")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
