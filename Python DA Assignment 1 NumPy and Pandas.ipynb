# Pandas Series

## Creating a Pandas Series

```python
marks=pd.Series([95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks
```

## Indexing and Slicing

```python
#Access the 1st rank using integer index position

print(marks.iloc[[0]])
```

```python
#Retrieve top 3 ranks using loc

marks.loc[['Rank1','Rank2','Rank3']]
```

```python
#Retrieve the 3rd rank using iloc

print(marks.iloc[[2]])
```

```python
#Boolean masking – Marks greater than 90

marks[marks>90]
```

## Manipulating Series

```python
#Modify the 1st rank mark to 100

marks.iloc[0]=100
marks
```

```python
# Remove the last rank

marks=marks.drop('Rank5')
marks
```

```python
#Calculate CGPA by dividing each mark by 10

CGPA=marks/10
CGPA
```

# Pandas DataFrame

## Creating a DataFrame

```python
Dict = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
print(Dict)
```

```python
transactions=pd.DataFrame(Dict)
transactions
```

## Data Exploration

```python
#Display the DataFrame

transactions
```

```python
#Display the first 5 rows using head()

transactions.head()
```

```python
#Display the last 5 rows using tail()

transactions.tail()
```

```python
#Check the shape

transactions.shape
```

```python
#Display column names

transactions.columns
```

```python
#Check data types

transactions.dtypes
```

```python
#Get complete basic information

transactions.info()
```

```python
#Display only ProductCategory and Amount

transactions[['ProductCategory','Amount']]
```

```python
#Retrieve the last 3 columns using iloc

transactions.iloc[:,-3:]
```

```python
#Filter rows where Region is North and Amount is greater than 200

transactions[(transactions['Region']=='North')&(transactions['Amount']>200)]
```

```python
#Find value counts for ProductCategory

transactions['ProductCategory'].value_counts()
```

```python
#Find unique values in Region

transactions['Region'].unique()
```

```python
#Group by Region and calculate mean Amount

transactions.groupby('Region')['Amount'].mean()
```

## Manipulating the DataFrame

```python
#Modify Amount for TransactionID 102 to 165

transactions.loc[transactions['TransactionID']==102,'Amount']=165
transactions
```

```python
#Add a new column Discount by calculating 10% of Amount

transactions['Discount']=transactions['Amount']*0.1
transactions
```

```python
#Remove the row with TransactionID 109

transactions=transactions.drop(transactions[transactions['TransactionID'] == 109].index)
transactions
```

```python
#Delete the Discount column

transactions=transactions.drop(['Discount'],axis=1)
transactions
```
